```python
import requests
import numpy as np
import pandas as pd

def get_past_24h_sequence(lat, lon):
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": lat,
        "longitude": lon,
        "past_days": 1,        # Requests yesterday's 24 hours
        "forecast_days": 0,    # We only want past data, not future forecast
        "hourly": ["shortwave_radiation", "temperature_2m", "relative_humidity_2m", "cloud_cover"],
        "timezone": "Asia/Singapore"
    }
    
    response = requests.get(url, params=params)
    data = response.json()
    
    # Convert JSON to a Pandas DataFrame
    df = pd.DataFrame(data["hourly"])
    
    # Grab exactly the last 24 rows
    df_24h = df.tail(24)
    
    # Extract as a NumPy array (shape: 24, 4) ready for your MinMaxScaler!
    feature_matrix = df_24h[["shortwave_radiation", "temperature_2m", "relative_humidity_2m", "cloud_cover"]].values
    
    return feature_matrix

# Try it out:
sequence = get_past_24h_sequence(3.1390, 101.6869)
print("Retrieved sequence shape:", sequence.shape)  # Output: (24, 4)
```

```python
import pandas as pd
import requests
import time
import json
import os

# --- setup ---
locations = pd.read_csv('locations_with_coords.csv')
os.makedirs('raw_nasa_power', exist_ok=True)

BASE_URL = "https://power.larc.nasa.gov/api/temporal/hourly/point"
PARAMS_TEMPLATE = {
    "parameters": "ALLSKY_SFC_SW_DWN",
    "community": "RE",
    "start": "20010101",   # corrected start date
    "end": "20241231",
    "format": "JSON"
}

failed = []
EXPECTED_RECORDS = 210384  # sanity check threshold

def fetch_station(row, retries=3, wait=5):
    lat, lon, place = row['lat'], row['lon'], row['place']
    safe_name = "".join(c if c.isalnum() else "_" for c in str(place))
    fname = f"raw_nasa_power/{safe_name}.json"

    # skip if already fetched (lets you resume after a crash)
    if os.path.exists(fname):
        return True

    params = PARAMS_TEMPLATE.copy()
    params.update({"latitude": lat, "longitude": lon})

    for attempt in range(1, retries + 1):
        try:
            resp = requests.get(BASE_URL, params=params, timeout=60)
            resp.raise_for_status()
            data = resp.json()

            values = data.get("properties", {}).get("parameter", {}).get("ALLSKY_SFC_SW_DWN", {})
            if len(values) < EXPECTED_RECORDS * 0.99:  # allow tiny tolerance
                raise ValueError(f"Incomplete: got {len(values)} records, expected ~{EXPECTED_RECORDS}")

            with open(fname, 'w') as f:
                json.dump(data, f)
            return True

        except Exception as e:
            print(f"  Attempt {attempt}/{retries} failed for {place}: {e}")
            time.sleep(wait)

    return False

# --- main loop ---
for idx, row in locations.iterrows():
    print(f"[{idx+1}/{len(locations)}] {row['place']}...", end=" ")
    success = fetch_station(row)
    print("OK" if success else "FAILED")
    if not success:
        failed.append(row['place'])
    time.sleep(1)  # gap between calls

print(f"\nDone. {len(failed)} stations failed after retries.")
if failed:
    pd.Series(failed, name="place").to_csv('failed_stations.csv', index=False)
    print(failed)
```

```python
import pandas as pd
import json
import os

RAW_DIR = "raw_nasa_power"
locations = pd.read_csv('locations_with_coords.csv')

output_file = "solar_radiation_intermediate.csv"
first_write = True

for idx, row in locations.iterrows():
    place = row['place']
    safe_name = "".join(c if c.isalnum() else "_" for c in str(place))
    fpath = os.path.join(RAW_DIR, f"{safe_name}.json")

    if not os.path.exists(fpath):
        print(f"Missing file for {place}, skipping")
        continue

    with open(fpath, 'r') as f:
        data = json.load(f)

    values = data["properties"]["parameter"]["ALLSKY_SFC_SW_DWN"]

    station_df = pd.DataFrame({
        "place": place,
        "datetime": pd.to_datetime(list(values.keys()), format="%Y%m%d%H"),
        "solar_radiation": list(values.values())
    })

    # append per station instead of holding everything in memory
    station_df.to_csv(output_file, mode='a', header=first_write, index=False)
    first_write = False

    print(f"[{idx+1}/{len(locations)}] {place}: {len(station_df)} rows written")

print("Done.")
```

```python
import pandas as pd
import gc

# --- load the smaller file fully (fits comfortably in RAM) ---
solar_df = pd.read_csv("solar_radiation_intermediate.csv")
solar_df['datetime'] = pd.to_datetime(solar_df['datetime'])
solar_df = solar_df.rename(columns={
    'datetime': 'datetime_hour',
    'solar_radiation': 'solar_radiation_nasa'
})

print(f"Solar reference data loaded: {len(solar_df):,} rows")

# --- process ground weather data in chunks ---
CHUNK_SIZE = 200_000
input_path = 'C:/Users/User/Downloads/solar_malaysia/full_weather.csv'
output_path = 'full_weather_merged.csv'

total_matched = 0
total_excluded = 0
first_write = True

reader = pd.read_csv(input_path, chunksize=CHUNK_SIZE)

for i, chunk in enumerate(reader):
    chunk['datetime'] = pd.to_datetime(chunk['datetime'])
    chunk['datetime_hour'] = chunk['datetime'].dt.round('h')

    merged = chunk.merge(
        solar_df,
        on=['place', 'datetime_hour'],
        how='left'
    )

    # fill the existing (empty) solar_radiation column with NASA values
    merged['solar_radiation'] = merged['solar_radiation_nasa']

    # flag rows with no NASA match (e.g. pre-2001) — exclude, don't dummy-fill
    matched_mask = merged['solar_radiation_nasa'].notna()
    excluded = (~matched_mask).sum()
    matched = matched_mask.sum()

    total_matched += matched
    total_excluded += excluded

    merged = merged[matched_mask].drop(columns=['datetime_hour', 'solar_radiation_nasa'])

    merged.to_csv(output_path, mode='a', header=first_write, index=False)
    first_write = False

    print(f"Chunk {i+1}: {matched:,} matched, {excluded:,} excluded (no NASA data)")

    del chunk, merged
    gc.collect()

print(f"\nDone. Total matched: {total_matched:,}, total excluded: {total_excluded:,}")
```

```python
import pandas as pd
import numpy as np
import requests
import json
import time
import glob
import os
import gc

# ==========================================
# STEP 1: CHUNKED LOADING TO PREVENT MEMORY ERROR
# ==========================================
print("=== STEP 1: Pass 1 - Auditing Stations in Chunks ===")

csv_file = 'full_weather_merged.csv'
chunk_size = 500_000

station_counts = {}
station_missing_temp = {}

# Pass 1: Compute missing temperature percentages using lightweight chunks
for chunk in pd.read_csv(csv_file, usecols=['place', 'temperature'], chunksize=chunk_size):
    for place, group in chunk.groupby('place'):
        station_counts[place] = station_counts.get(place, 0) + len(group)
        station_missing_temp[place] = station_missing_temp.get(place, 0) + group['temperature'].isnull().sum()

# Identify operational stations (< 20% missing temperature)
valid_stations = [
    place for place, total in station_counts.items() 
    if total > 0 and (station_missing_temp.get(place, 0) / total) < 0.20
]

print(f"Identified {len(valid_stations)} operational stations out of {len(station_counts)} total stations.")

print("\n=== Pass 2: Memory-Efficient Loading & Downcasting ===")

# Columns to load (ignoring high-missing sparse columns)
cols_to_load = [
    'datetime', 'place', 'city', 'state', 'temperature', 'pressure', 
    'dew_point', 'humidity', 'wind_speed', 'wind_chill', 'uv_index', 
    'solar_radiation', 'precipitation_rate', 'precipitation_total'
]

# Downcast data types to save ~70% RAM
dtypes = {
    'temperature': 'float32',
    'pressure': 'float32',
    'dew_point': 'float32',
    'humidity': 'float32',
    'wind_speed': 'float32',
    'wind_chill': 'float32',
    'uv_index': 'float32',
    'solar_radiation': 'float32',
    'precipitation_rate': 'float32',
    'precipitation_total': 'float32',
    'place': 'category',
    'city': 'category',
    'state': 'category'
}

chunks = []
for chunk in pd.read_csv(csv_file, usecols=lambda c: c in cols_to_load, dtype=dtypes, chunksize=chunk_size):
    filtered_chunk = chunk[chunk['place'].isin(valid_stations)]
    if not filtered_chunk.empty:
        chunks.append(filtered_chunk)

df_filtered = pd.concat(chunks, ignore_index=True)

# Free up unneeded memory chunks
del chunks
gc.collect()

print(f"Successfully loaded clean filtered dataset! Shape: {df_filtered.shape}")

# ==========================================
# STEP 2: FETCH NASA UV DATA FOR ACTIVE STATIONS
# ==========================================
print("\n=== STEP 2: Fetching NASA UV Index Data ===")

locations = pd.read_csv('locations_with_coords.csv')
lat_col = 'lat' if 'lat' in locations.columns else 'latitude'
lon_col = 'lon' if 'lon' in locations.columns else 'longitude'

locations_active = locations[locations['place'].isin(valid_stations)].copy()
os.makedirs('raw_nasa_uv', exist_ok=True)

BASE_URL = "https://power.larc.nasa.gov/api/temporal/hourly/point"
PARAMS_TEMPLATE = {
    "parameters": "ALLSKY_SFC_UV_INDEX",
    "community": "RE",
    "start": "20010101",
    "end": "20241231",
    "format": "JSON"
}

EXPECTED_RECORDS = 210384

def fetch_uv_station(row, retries=3, wait=5):
    lat, lon, place = row[lat_col], row[lon_col], row['place']
    safe_name = "".join(c if c.isalnum() else "_" for c in str(place))
    fname = f"raw_nasa_uv/{safe_name}_uv.json"

    if os.path.exists(fname):
        return True

    params = PARAMS_TEMPLATE.copy()
    params.update({"latitude": lat, "longitude": lon})

    for attempt in range(1, retries + 1):
        try:
            resp = requests.get(BASE_URL, params=params, timeout=60)
            resp.raise_for_status()
            data = resp.json()

            values = data.get("properties", {}).get("parameter", {}).get("ALLSKY_SFC_UV_INDEX", {})
            if len(values) < EXPECTED_RECORDS * 0.99:
                raise ValueError(f"Incomplete UV records: got {len(values)}")

            with open(fname, 'w') as f:
                json.dump(data, f)
            return True

        except Exception as e:
            print(f"  Attempt {attempt} failed for {place}: {e}")
            time.sleep(wait)

    return False

for idx, row in locations_active.reset_index(drop=True).iterrows():
    print(f"[{idx+1}/{len(locations_active)}] Fetching UV for {row['place']}...", end=" ")
    success = fetch_uv_station(row)
    print("OK" if success else "FAILED")
    time.sleep(0.5)

# ==========================================
# STEP 3: PARSE JSONs & MERGE NASA UV
# ==========================================
print("\n=== STEP 3: Merging NASA UV Data into Dataset ===")

safename_to_place = {
    "".join(c if c.isalnum() else "_" for c in str(p)): str(p)
    for p in locations_active['place'].unique()
}

uv_records = []
json_files = glob.glob("raw_nasa_uv/*_uv.json")

for filepath in json_files:
    safe_name = os.path.basename(filepath).replace('_uv.json', '')
    original_place = safename_to_place.get(safe_name, None)
    
    if original_place is None:
        continue

    with open(filepath, 'r') as f:
        data = json.load(f)
    
    uv_dict = data['properties']['parameter']['ALLSKY_SFC_UV_INDEX']
    temp_df = pd.DataFrame(list(uv_dict.items()), columns=['raw_time', 'nasa_uv'])
    temp_df['datetime'] = pd.to_datetime(temp_df['raw_time'], format='%Y%m%d%H')
    temp_df['place'] = original_place
    
    # NASA missing value sentinel
    temp_df['nasa_uv'] = temp_df['nasa_uv'].replace(-999, np.nan).astype('float32')
    uv_records.append(temp_df[['datetime', 'place', 'nasa_uv']])

nasa_uv_df = pd.concat(uv_records, ignore_index=True)

# Merge back with main dataset
df_filtered['datetime'] = pd.to_datetime(df_filtered['datetime'])
df_filtered['place'] = df_filtered['place'].astype(str)
nasa_uv_df['place'] = nasa_uv_df['place'].astype(str)

df_merged = df_filtered.merge(nasa_uv_df, on=['datetime', 'place'], how='left')

if 'uv_index' in df_merged.columns:
    df_merged['uv_index'] = df_merged['uv_index'].fillna(df_merged['nasa_uv'])
else:
    df_merged['uv_index'] = df_merged['nasa_uv']

df_merged.drop(columns=['nasa_uv'], inplace=True)

# ==========================================
# STEP 4: INTERPOLATION & FINAL CLEANUP
# ==========================================
print("\n=== STEP 4: Interpolating Minor Gaps & Exporting ===")

minor_cols = ['humidity', 'wind_speed', 'wind_chill', 'uv_index', 'temperature', 'pressure', 'dew_point']
for col in minor_cols:
    if col in df_merged.columns:
        df_merged[col] = df_merged.groupby('place')[col].transform(
            lambda x: x.interpolate(method='linear').ffill().bfill()
        )

if 'precipitation_rate' in df_merged.columns:
    df_merged['precipitation_rate'] = df_merged['precipitation_rate'].fillna(0.0)
if 'precipitation_total' in df_merged.columns:
    df_merged['precipitation_total'] = df_merged['precipitation_total'].fillna(0.0)

print("\nFinal Check - Remaining Missing Values:")
print(df_merged.isnull().sum())

# Export clean, uncorrupted dataset
output_filename = 'final_weather_cleaned.csv'
df_merged.to_csv(output_filename, index=False)
print(f"\nSUCCESS! Clean dataset saved to '{output_filename}' ({len(df_merged):,} rows).")
```

```python
import pandas as pd
import numpy as np
import json
import glob
import os
import gc

# 1. Load and parse all cached UV JSON files into one lookup dataframe
uv_records = []
for filepath in glob.glob('raw_nasa_uv/*_uv.json'):
    with open(filepath, 'r') as f:
        data = json.load(f)
    uv_dict = data.get('properties', {}).get('parameter', {}).get('ALLSKY_SFC_UV_INDEX', {})
    if not uv_dict:
        continue
    temp_df = pd.DataFrame(list(uv_dict.items()), columns=['raw_time', 'nasa_uv'])
    temp_df['datetime'] = pd.to_datetime(temp_df['raw_time'], format='%Y%m%d%H')
    safe_name = os.path.basename(filepath).replace('_uv.json', '')
    temp_df['safe_place'] = safe_name
    temp_df['nasa_uv'] = temp_df['nasa_uv'].replace(-999, np.nan).astype('float32')
    uv_records.append(temp_df[['datetime', 'safe_place', 'nasa_uv']])

nasa_uv_df = pd.concat(uv_records, ignore_index=True)
print(f"NASA UV records loaded: {len(nasa_uv_df):,}")

# 2. Map safe filenames back to real station names
audit = pd.read_csv('station_audit.csv')
valid_stations = list(audit[audit['temperature_missing_pct'] < 20]['place'])
place_to_safe = {p: "".join(c if c.isalnum() else "_" for c in str(p)) for p in valid_stations}
safe_to_place = {v: k for k, v in place_to_safe.items()}
nasa_uv_df['place'] = nasa_uv_df['safe_place'].map(safe_to_place)
nasa_uv_df = nasa_uv_df.dropna(subset=['place']).drop(columns=['safe_place'])

# 3. Chunked merge into the main dataset
CHUNK_SIZE = 300_000
input_path = 'weather_columns_dropped.csv'
output_path = 'weather_uv_merged.csv'

dtypes = {
    'temperature': 'float32', 'pressure': 'float32', 'dew_point': 'float32',
    'humidity': 'float32', 'wind_speed': 'float32', 'wind_chill': 'float32',
    'uv_index': 'float32', 'solar_radiation': 'float32',
    'precipitation_rate': 'float32', 'precipitation_total': 'float32',
}

first_write = True
total_rows = 0
uv_filled_count = 0

for i, chunk in enumerate(pd.read_csv(input_path, dtype=dtypes, chunksize=CHUNK_SIZE)):
    chunk['datetime'] = pd.to_datetime(chunk['datetime'])
    before_missing = chunk['uv_index'].isnull().sum()

    chunk = chunk.merge(nasa_uv_df, on=['datetime', 'place'], how='left')
    # only fill where ground uv_index is genuinely missing — never overwrite real readings
    chunk['uv_index'] = chunk['uv_index'].fillna(chunk['nasa_uv'])
    chunk = chunk.drop(columns=['nasa_uv'])

    after_missing = chunk['uv_index'].isnull().sum()
    uv_filled_count += (before_missing - after_missing)

    chunk.to_csv(output_path, mode='a', header=first_write, index=False)
    first_write = False
    total_rows += len(chunk)

    if (i + 1) % 10 == 0:
        print(f"Processed chunk {i+1}, running total: {total_rows:,} rows")

    del chunk
    gc.collect()

print(f"\nDone. Total rows: {total_rows:,}")
print(f"UV values filled from NASA: {uv_filled_count:,}")
```

```python
import pandas as pd
import requests
import json
import time
import os

# Load the coordinates for your 71 valid stations
locations = pd.read_csv('locations_with_coords.csv')
lat_col = 'lat' if 'lat' in locations.columns else 'latitude'
lon_col = 'lon' if 'lon' in locations.columns else 'longitude'

audit = pd.read_csv('station_audit.csv')
valid_stations = list(audit[audit['temperature_missing_pct'] < 20]['place'])
locations_active = locations[locations['place'].isin(valid_stations)].copy()
print(f"Re-fetching UV for {len(locations_active)} stations")

# Save to a FRESH folder so we don't mix old and new results
os.makedirs('raw_nasa_uv_v2', exist_ok=True)

BASE_URL = "https://power.larc.nasa.gov/api/temporal/hourly/point"
PARAMS_TEMPLATE = {
    "parameters": "ALLSKY_SFC_UV_INDEX",
    "community": "RE",
    "start": "20010101",
    "end": "20241231",
    "format": "JSON"
}
EXPECTED_RECORDS = 210384

def fetch_uv_station(row, retries=3, wait=5):
    lat, lon, place = row[lat_col], row[lon_col], row['place']
    safe_name = "".join(c if c.isalnum() else "_" for c in str(place))
    fname = f"raw_nasa_uv_v2/{safe_name}_uv.json"

    params = PARAMS_TEMPLATE.copy()
    params.update({"latitude": lat, "longitude": lon})

    for attempt in range(1, retries + 1):
        try:
            resp = requests.get(BASE_URL, params=params, timeout=60)
            resp.raise_for_status()
            data = resp.json()
            values = data.get("properties", {}).get("parameter", {}).get("ALLSKY_SFC_UV_INDEX", {})
            if len(values) < EXPECTED_RECORDS * 0.99:
                raise ValueError(f"Incomplete UV records: got {len(values)}")
            with open(fname, 'w') as f:
                json.dump(data, f)
            return True
        except Exception as e:
            print(f"  Attempt {attempt} failed for {place}: {e}")
            time.sleep(wait)
    return False

results = []
for idx, row in locations_active.reset_index(drop=True).iterrows():
    print(f"[{idx+1}/{len(locations_active)}] Fetching UV for {row['place']}...", end=" ")
    success = fetch_uv_station(row)
    results.append((row['place'], success))
    print("OK" if success else "FAILED")
    time.sleep(0.5)

failed = [p for p, ok in results if not ok]
print(f"\nDone. {len(results) - len(failed)} succeeded, {len(failed)} failed.")
if failed:
    print("Failed stations:", failed)
```

```python
import pandas as pd

dtypes = {
    'temperature': 'float32', 'pressure': 'float32', 'dew_point': 'float32',
    'humidity': 'float32', 'wind_speed': 'float32', 'wind_chill': 'float32',
    'uv_index': 'float32', 'solar_radiation': 'float32',
}
df = pd.read_csv('weather_precip_dropped.csv', dtype=dtypes)
df['datetime'] = pd.to_datetime(df['datetime'])
print(f"Rows before trim: {len(df):,}")

# "Real reading" = temperature is present (our most reliable core sensor)
core_col = 'temperature'
trimmed = []
for place, g in df.groupby('place'):
    valid = g[g[core_col].notna()]
    if valid.empty:
        continue
    start, end = valid['datetime'].min(), valid['datetime'].max()
    trimmed.append(g[(g['datetime'] >= start) & (g['datetime'] <= end)])

df_trimmed = pd.concat(trimmed, ignore_index=True)
print(f"Rows after trim: {len(df_trimmed):,}")
print(f"Rows removed (pre/post-installation NaN edges): {len(df) - len(df_trimmed):,}")

df_trimmed.to_csv('weather_trimmed.csv', index=False)
print("Saved to weather_trimmed.csv")
```

```python
import pandas as pd

dtypes = {
    'temperature': 'float32', 'pressure': 'float32', 'dew_point': 'float32',
    'humidity': 'float32', 'wind_speed': 'float32', 'wind_chill': 'float32',
    'uv_index': 'float32', 'solar_radiation': 'float32',
}
df = pd.read_csv('weather_final_clean.csv', dtype=dtypes)
df['datetime'] = pd.to_datetime(df['datetime'])
df = df.sort_values(['place', 'datetime']).reset_index(drop=True)

print(f"Total rows: {len(df):,}")

train_parts, val_parts, test_parts = [], [], []

for place, g in df.groupby('place'):
    n = len(g)
    train_end = int(n * 0.70)
    val_end = int(n * 0.85)  # 70% + 15%

    train_parts.append(g.iloc[:train_end])
    val_parts.append(g.iloc[train_end:val_end])
    test_parts.append(g.iloc[val_end:])

df_train = pd.concat(train_parts, ignore_index=True)
df_val = pd.concat(val_parts, ignore_index=True)
df_test = pd.concat(test_parts, ignore_index=True)

print(f"Train: {len(df_train):,} ({len(df_train)/len(df)*100:.1f}%)")
print(f"Val:   {len(df_val):,} ({len(df_val)/len(df)*100:.1f}%)")
print(f"Test:  {len(df_test):,} ({len(df_test)/len(df)*100:.1f}%)")

# Sanity check: confirm chronological order is respected per station (no leakage)
for place in df['place'].unique()[:3]:  # spot-check a few
    tr_max = df_train[df_train['place'] == place]['datetime'].max()
    va_min = df_val[df_val['place'] == place]['datetime'].min()
    va_max = df_val[df_val['place'] == place]['datetime'].max()
    te_min = df_test[df_test['place'] == place]['datetime'].min()
    print(f"\n{place}: train ends {tr_max}, val {va_min}->{va_max}, test starts {te_min}")
    assert tr_max < va_min, "LEAKAGE: train overlaps val!"
    assert va_max < te_min, "LEAKAGE: val overlaps test!"

df_train.to_csv('train_raw.csv', index=False)
df_val.to_csv('val_raw.csv', index=False)
df_test.to_csv('test_raw.csv', index=False)
print("\nSaved train_raw.csv, val_raw.csv, test_raw.csv")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import pickle

dtypes = {
    'temperature': 'float32', 'pressure': 'float32', 'dew_point': 'float32',
    'humidity': 'float32', 'wind_speed': 'float32', 'wind_chill': 'float32',
    'uv_index': 'float32', 'solar_radiation': 'float32',
}
train = pd.read_csv('train_raw.csv', dtype=dtypes)
val = pd.read_csv('val_raw.csv', dtype=dtypes)
test = pd.read_csv('test_raw.csv', dtype=dtypes)

for d in [train, val, test]:
    d['datetime'] = pd.to_datetime(d['datetime'])

feature_cols = ['temperature', 'pressure', 'dew_point', 'humidity',
                 'wind_speed', 'wind_chill', 'uv_index', 'solar_radiation']

scalers = {}

train_scaled_parts, val_scaled_parts, test_scaled_parts = [], [], []

for place in train['place'].unique():
    tr_g = train[train['place'] == place].copy()
    va_g = val[val['place'] == place].copy()
    te_g = test[test['place'] == place].copy()

    scaler = StandardScaler()
    tr_g[feature_cols] = scaler.fit_transform(tr_g[feature_cols])  # fit ONLY on train

    if not va_g.empty:
        va_g[feature_cols] = scaler.transform(va_g[feature_cols])  # apply, don't refit
    if not te_g.empty:
        te_g[feature_cols] = scaler.transform(te_g[feature_cols])  # apply, don't refit

    scalers[place] = scaler
    train_scaled_parts.append(tr_g)
    val_scaled_parts.append(va_g)
    test_scaled_parts.append(te_g)

train_scaled = pd.concat(train_scaled_parts, ignore_index=True)
val_scaled = pd.concat(val_scaled_parts, ignore_index=True)
test_scaled = pd.concat(test_scaled_parts, ignore_index=True)

print(f"Train scaled: {len(train_scaled):,} rows")
print(f"Val scaled:   {len(val_scaled):,} rows")
print(f"Test scaled:  {len(test_scaled):,} rows")

# Sanity check: scaled train features should be ~mean 0, std 1
print("\nTrain feature stats after scaling (should be ~0 mean, ~1 std):")
print(train_scaled[feature_cols].describe().loc[['mean', 'std']])

train_scaled.to_csv('train_scaled.csv', index=False)
val_scaled.to_csv('val_scaled.csv', index=False)
test_scaled.to_csv('test_scaled.csv', index=False)

with open('station_scalers.pkl', 'wb') as f:
    pickle.dump(scalers, f)

print("\nSaved train_scaled.csv, val_scaled.csv, test_scaled.csv, station_scalers.pkl")
```

```python
import pandas as pd
import numpy as np

def build_sequences(df, feature_cols, target_col='solar_radiation', window_hours=24):
    """
    Builds sliding-window sequences per station, skipping any window
    where timestamps aren't perfectly consecutive hourly steps (i.e.
    respects gaps from excluded rows rather than silently bridging them).
    """
    X_list, y_list, meta_list = [], [], []

    for place, g in df.groupby('place'):
        g = g.sort_values('datetime').reset_index(drop=True)
        values = g[feature_cols].values
        targets = g[target_col].values
        times = g['datetime'].values

        for i in range(len(g) - window_hours):
            window_times = times[i : i + window_hours + 1]  # +1 to include the target hour
            # Check every consecutive pair is exactly 1 hour apart
            diffs = np.diff(window_times).astype('timedelta64[h]').astype(int)
            if not np.all(diffs == 1):
                continue  # gap inside this window — skip it, don't fabricate continuity

            X_list.append(values[i : i + window_hours])
            y_list.append(targets[i + window_hours])
            meta_list.append((place, times[i + window_hours]))

    X = np.array(X_list)
    y = np.array(y_list)
    return X, y, meta_list

# Quick test run on a small window length first, to check it behaves before scaling to full data
train = pd.read_csv('train_scaled.csv', dtype={'place': str})
train['datetime'] = pd.to_datetime(train['datetime'])

feature_cols = ['temperature', 'pressure', 'dew_point', 'humidity',
                 'wind_speed', 'wind_chill', 'uv_index', 'solar_radiation']

X_train_24, y_train_24, meta_24 = build_sequences(train, feature_cols, window_hours=24)
print(f"24h windows built: {X_train_24.shape[0]:,}")
print(f"X shape: {X_train_24.shape}  (samples, timesteps, features)")
print(f"y shape: {y_train_24.shape}")
```

```python
import pandas as pd
import numpy as np
import gc

def build_sequences(df, feature_cols, target_col='solar_radiation', window_hours=24):
    X_list, y_list, meta_list = [], [], []
    for place, g in df.groupby('place'):
        g = g.sort_values('datetime').reset_index(drop=True)
        values = g[feature_cols].values
        targets = g[target_col].values
        times = g['datetime'].values

        for i in range(len(g) - window_hours):
            window_times = times[i : i + window_hours + 1]
            diffs = np.diff(window_times).astype('timedelta64[h]').astype(int)
            if not np.all(diffs == 1):
                continue
            X_list.append(values[i : i + window_hours])
            y_list.append(targets[i + window_hours])
            meta_list.append((place, times[i + window_hours]))

    X = np.array(X_list, dtype='float32')
    y = np.array(y_list, dtype='float32')
    return X, y, meta_list

feature_cols = ['temperature', 'pressure', 'dew_point', 'humidity',
                 'wind_speed', 'wind_chill', 'uv_index', 'solar_radiation']

dtypes = {c: 'float32' for c in feature_cols}
dtypes['place'] = str

for split_name in ['train', 'val', 'test']:
    df = pd.read_csv(f'{split_name}_scaled.csv', dtype=dtypes)
    df['datetime'] = pd.to_datetime(df['datetime'])

    for window_hours in [24, 48, 72]:
        print(f"Building {split_name} / {window_hours}h windows...")
        X, y, meta = build_sequences(df, feature_cols, window_hours=window_hours)
        print(f"  -> {X.shape[0]:,} sequences, shape {X.shape}")

        np.save(f'{split_name}_X_{window_hours}h.npy', X)
        np.save(f'{split_name}_y_{window_hours}h.npy', y)

        del X, y, meta
        gc.collect()

    del df
    gc.collect()

print("\nAll window lengths built for train/val/test.")
```

```python
import pandas as pd
import numpy as np

dtypes = {c: 'float32' for c in ['temperature', 'pressure', 'dew_point', 'humidity',
                                   'wind_speed', 'wind_chill', 'uv_index', 'solar_radiation']}
dtypes['place'] = str

def build_xgb_features(input_path, output_path):
    df = pd.read_csv(input_path, dtype=dtypes)
    df['datetime'] = pd.to_datetime(df['datetime'])
    df = df.sort_values(['place', 'datetime']).reset_index(drop=True)

    all_parts = []
    for place, g in df.groupby('place'):
        g = g.set_index('datetime').sort_index()

        # Reindex to a full hourly grid — any missing hour becomes a real NaN row,
        # so lag/rolling calculations can't silently skip across a gap
        full_range = pd.date_range(g.index.min(), g.index.max(), freq='h')
        g = g.reindex(full_range)
        g['place'] = place  # reindex wipes non-numeric cols with NaN, restore this

        # Lag features
        for lag in [1, 2, 3, 24]:
            g[f'solar_radiation_lag_{lag}'] = g['solar_radiation'].shift(lag)
            g[f'temperature_lag_{lag}'] = g['temperature'].shift(lag)

        # Rolling stats (time-aware — respects the real hourly index, gaps included)
        g['solar_radiation_roll_mean_3h'] = g['solar_radiation'].rolling('3h').mean()
        g['solar_radiation_roll_std_3h'] = g['solar_radiation'].rolling('3h').std()
        g['solar_radiation_roll_mean_24h'] = g['solar_radiation'].rolling('24h').mean()
        g['solar_radiation_roll_std_24h'] = g['solar_radiation'].rolling('24h').std()
        g['temperature_roll_mean_3h'] = g['temperature'].rolling('3h').mean()
        g['temperature_roll_std_3h'] = g['temperature'].rolling('3h').std()
        g['temperature_roll_mean_24h'] = g['temperature'].rolling('24h').mean()
        g['temperature_roll_std_24h'] = g['temperature'].rolling('24h').std()

        g = g.reset_index().rename(columns={'index': 'datetime'})
        all_parts.append(g)

    result = pd.concat(all_parts, ignore_index=True)

    # Calendar features
    result['hour_of_day'] = result['datetime'].dt.hour
    result['month_of_year'] = result['datetime'].dt.month
    # Malaysia: Northeast monsoon ~Nov-Mar, Southwest monsoon ~May-Sep (roughly)
    result['is_northeast_monsoon'] = result['month_of_year'].isin([11, 12, 1, 2, 3]).astype(int)
    result['is_southwest_monsoon'] = result['month_of_year'].isin([5, 6, 7, 8, 9]).astype(int)

    # Drop rows where lag/rolling features couldn't be computed (start of each station's timeline,
    # or right after a real gap) — XGBoost needs complete feature rows, same exclude-don't-fill principle
    feature_cols = [c for c in result.columns if 'lag' in c or 'roll' in c]
    before = len(result)
    result = result.dropna(subset=feature_cols)
    print(f"{output_path}: {before:,} -> {len(result):,} rows "
          f"(dropped {before - len(result):,} with incomplete lag/rolling history)")

    result.to_csv(output_path, index=False)
    return result

for split in ['train', 'val', 'test']:
    build_xgb_features(f'{split}_scaled.csv', f'{split}_xgb_features.csv')
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import pickle

dtypes = {c: 'float32' for c in ['temperature','pressure','dew_point','humidity',
                                  'wind_speed','wind_chill','uv_index','solar_radiation']}
train = pd.read_csv('train_raw_v2.csv', dtype=dtypes)
val = pd.read_csv('val_raw_v2.csv', dtype=dtypes)
test = pd.read_csv('test_raw_v2.csv', dtype=dtypes)
for d in [train, val, test]:
    d['datetime'] = pd.to_datetime(d['datetime'])

complete_cols = ['temperature', 'pressure', 'dew_point', 'humidity',
                  'wind_speed', 'wind_chill', 'solar_radiation']  # no NaN in these
uv_col = 'uv_index'  # has NaN — handle separately

scalers = {}  # will store {place: {'main': scaler, 'uv_mean':..., 'uv_std':...}}
train_parts, val_parts, test_parts = [], [], []

for place in train['place'].unique():
    tr_g = train[train['place'] == place].copy()
    va_g = val[val['place'] == place].copy()
    te_g = test[test['place'] == place].copy()

    # Scale the 7 complete columns normally
    scaler = StandardScaler()
    tr_g[complete_cols] = scaler.fit_transform(tr_g[complete_cols])
    if not va_g.empty:
        va_g[complete_cols] = scaler.transform(va_g[complete_cols])
    if not te_g.empty:
        te_g[complete_cols] = scaler.transform(te_g[complete_cols])

    # Scale uv_index manually using train's mean/std, NaN stays NaN automatically
    uv_mean = tr_g[uv_col].mean()   # pandas .mean() ignores NaN by default
    uv_std = tr_g[uv_col].std()
    tr_g[uv_col] = (tr_g[uv_col] - uv_mean) / uv_std
    if not va_g.empty:
        va_g[uv_col] = (va_g[uv_col] - uv_mean) / uv_std
    if not te_g.empty:
        te_g[uv_col] = (te_g[uv_col] - uv_mean) / uv_std

    scalers[place] = {'scaler': scaler, 'uv_mean': uv_mean, 'uv_std': uv_std}
    train_parts.append(tr_g)
    val_parts.append(va_g)
    test_parts.append(te_g)

train_scaled = pd.concat(train_parts, ignore_index=True)
val_scaled = pd.concat(val_parts, ignore_index=True)
test_scaled = pd.concat(test_parts, ignore_index=True)

print(f"Train: {len(train_scaled):,}, uv_index NaN: {train_scaled['uv_index'].isnull().sum():,}")
print(f"Val:   {len(val_scaled):,}, uv_index NaN: {val_scaled['uv_index'].isnull().sum():,}")
print(f"Test:  {len(test_scaled):,}, uv_index NaN: {test_scaled['uv_index'].isnull().sum():,}")
print("\nComplete-column stats (should be ~0 mean, ~1 std):")
print(train_scaled[complete_cols].describe().loc[['mean', 'std']])

train_scaled.to_csv('train_scaled_v2.csv', index=False)
val_scaled.to_csv('val_scaled_v2.csv', index=False)
test_scaled.to_csv('test_scaled_v2.csv', index=False)
with open('station_scalers_v2.pkl', 'wb') as f:
    pickle.dump(scalers, f)
print("\nSaved train_scaled_v2.csv, val_scaled_v2.csv, test_scaled_v2.csv, station_scalers_v2.pkl")
```

```python
import pandas as pd
import numpy as np
import gc

def build_sequences(df, feature_cols, target_col='solar_radiation', window_hours=24):
    X_list, y_list, meta_list = [], [], []
    for place, g in df.groupby('place'):
        g = g.sort_values('datetime').reset_index(drop=True)
        values = g[feature_cols].values
        targets = g[target_col].values
        times = g['datetime'].values
        for i in range(len(g) - window_hours):
            window_times = times[i : i + window_hours + 1]
            diffs = np.diff(window_times).astype('timedelta64[h]').astype(int)
            if not np.all(diffs == 1):
                continue
            X_list.append(values[i : i + window_hours])
            y_list.append(targets[i + window_hours])
            meta_list.append((place, times[i + window_hours]))
    X = np.array(X_list, dtype='float32')
    y = np.array(y_list, dtype='float32')
    return X, y, meta_list

# 7 features — uv_index excluded for sequence models
seq_feature_cols = ['temperature', 'pressure', 'dew_point', 'humidity',
                     'wind_speed', 'wind_chill', 'solar_radiation']

dtypes = {c: 'float32' for c in seq_feature_cols + ['uv_index']}
dtypes['place'] = str

for split_name in ['train', 'val', 'test']:
    df = pd.read_csv(f'{split_name}_scaled_v2.csv', dtype=dtypes)
    df['datetime'] = pd.to_datetime(df['datetime'])

    for window_hours in [24, 48, 72]:
        print(f"Building {split_name} / {window_hours}h windows (7 features)...")
        X, y, meta = build_sequences(df, seq_feature_cols, window_hours=window_hours)
        print(f"  -> {X.shape[0]:,} sequences, shape {X.shape}")
        np.save(f'{split_name}_X_{window_hours}h_v2.npy', X)
        np.save(f'{split_name}_y_{window_hours}h_v2.npy', y)
        del X, y, meta
        gc.collect()
    del df
    gc.collect()

print("\nAll v2 window lengths built (7 features, no uv_index).")
```

```python
import pandas as pd
import numpy as np

dtypes = {c: 'float32' for c in ['temperature', 'pressure', 'dew_point', 'humidity',
                                   'wind_speed', 'wind_chill', 'uv_index', 'solar_radiation']}
dtypes['place'] = str

def build_xgb_features(input_path, output_path):
    df = pd.read_csv(input_path, dtype=dtypes)
    df['datetime'] = pd.to_datetime(df['datetime'])
    df = df.sort_values(['place', 'datetime']).reset_index(drop=True)

    all_parts = []
    for place, g in df.groupby('place'):
        g = g.set_index('datetime').sort_index()
        full_range = pd.date_range(g.index.min(), g.index.max(), freq='h')
        g = g.reindex(full_range)
        g['place'] = place

        for lag in [1, 2, 3, 24]:
            g[f'solar_radiation_lag_{lag}'] = g['solar_radiation'].shift(lag)
            g[f'temperature_lag_{lag}'] = g['temperature'].shift(lag)
            g[f'uv_index_lag_{lag}'] = g['uv_index'].shift(lag)

        # FIX: shift(1) before rolling — window only sees strictly past hours,
        # never the current row's own target value (prevents leakage)
        solar_past = g['solar_radiation'].shift(1)
        temp_past = g['temperature'].shift(1)

        g['solar_radiation_roll_mean_3h'] = solar_past.rolling('3h').mean()
        g['solar_radiation_roll_std_3h'] = solar_past.rolling('3h').std()
        g['solar_radiation_roll_mean_24h'] = solar_past.rolling('24h').mean()
        g['solar_radiation_roll_std_24h'] = solar_past.rolling('24h').std()
        g['temperature_roll_mean_3h'] = temp_past.rolling('3h').mean()
        g['temperature_roll_std_3h'] = temp_past.rolling('3h').std()
        g['temperature_roll_mean_24h'] = temp_past.rolling('24h').mean()
        g['temperature_roll_std_24h'] = temp_past.rolling('24h').std()

        g = g.reset_index().rename(columns={'index': 'datetime'})
        all_parts.append(g)

    result = pd.concat(all_parts, ignore_index=True)

    result['hour_of_day'] = result['datetime'].dt.hour
    result['month_of_year'] = result['datetime'].dt.month
    result['is_northeast_monsoon'] = result['month_of_year'].isin([11, 12, 1, 2, 3]).astype(int)
    result['is_southwest_monsoon'] = result['month_of_year'].isin([5, 6, 7, 8, 9]).astype(int)

    required_cols = ['solar_radiation',
                      'solar_radiation_lag_1', 'solar_radiation_lag_2',
                      'solar_radiation_lag_3', 'solar_radiation_lag_24',
                      'solar_radiation_roll_mean_3h', 'solar_radiation_roll_mean_24h']

    before = len(result)
    result = result.dropna(subset=required_cols)
    print(f"{output_path}: {before:,} -> {len(result):,} rows "
          f"(dropped {before - len(result):,} missing target/target-lag history)")
    print(f"  Remaining NaN in uv_index-related columns (expected): "
          f"{result[[c for c in result.columns if 'uv' in c]].isnull().sum().sum():,}")

    result.to_csv(output_path, index=False)
    return result

for split in ['train', 'val', 'test']:
    build_xgb_features(f'{split}_scaled_v2.csv', f'{split}_xgb_features_v2.csv')
```
